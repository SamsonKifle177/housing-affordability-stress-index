# 04 SQL Database Setup

## Objective

Load the processed housing affordability dataset into a local SQLite database for SQL analysis.

## Input

- `data/processed/housing_macro_with_stress_score.csv`

## Output

- `housing_affordability.db`

## Tables Created

- `housing_macro_country_year`

```python
import pandas as pd
import sqlite3
from pathlib import Path

PROCESSED_DIR = Path("../data/processed")
DB_PATH = Path("../housing_affordability.db")

print("Setup complete")
```

```python
df = pd.read_csv(PROCESSED_DIR / "housing_macro_with_stress_score.csv")

print("Shape:", df.shape)
df.head()
```

```python
conn = sqlite3.connect(DB_PATH)

df.to_sql(
    "housing_macro_country_year",
    conn,
    if_exists="replace",
    index=False
)

print("Table loaded successfully")
```

```python
query = """
SELECT
    country_name,
    year,
    housing_affordability_stress_score,
    housing_stress_category
FROM housing_macro_country_year
ORDER BY housing_affordability_stress_score DESC
LIMIT 10;
"""

pd.read_sql_query(query, conn)
```

```python
conn.close()
print("Database connection closed")
```
